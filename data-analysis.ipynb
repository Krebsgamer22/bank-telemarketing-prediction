# Univariate Analyse des Bank Marketing Datasets

Dieses Notebook führt eine umfassende univariate Analyse der Daten aus der `bank-full.csv` durch. Wir analysieren jede Variable einzeln, um ihre Verteilung, Statistiken und potenzielle Erkenntnisse zu verstehen.

## 1. Daten laden und Überblick

Laden wir die Daten und verschaffen uns einen ersten Überblick über die Struktur und die grundlegenden Statistiken.

```python
# Daten laden
df = pd.read_csv('bank-full.csv', sep=';')

# Grundlegender Überblick
print("Form des Datensatzes:", df.shape)
print("\nSpaltennamen:")
print(df.columns.tolist())
print("\nDatentypen:")
print(df.dtypes)
print("\nErste 5 Zeilen:")
display(df.head())
print("\nZusammenfassung der numerischen Variablen:")
display(df.describe())
print("\nZusammenfassung der kategorischen Variablen:")
display(df.describe(include=['object']))
print("\nFehlende Werte:")
print(df.isnull().sum())
```

## 2. Analyse numerischer Variablen

Wir analysieren die numerischen Variablen: age, balance, day, duration, campaign, pdays, previous.

Für jede Variable berechnen wir:
- Grundlegende Statistiken
- Verteilungsmaße (Schiefe, Kurtosis)
- Visualisierungen (Histogramme, Boxplots)

```python
# Funktion für univariate Analyse numerischer Variablen
def analyze_numeric_variable(df, column):
    print(f"\n=== Analyse von {column} ===")
    print(f"Anzahl der Werte: {len(df[column])}")
    print(f"Einzigartige Werte: {df[column].nunique()}")
    print(f"Fehlende Werte: {df[column].isnull().sum()}")
    print(f"Mittelwert: {df[column].mean():.2f}")
    print(f"Median: {df[column].median():.2f}")
    print(f"Modus: {df[column].mode().iloc[0]}")
    print(f"Standardabweichung: {df[column].std():.2f}")
    print(f"Minimum: {df[column].min()}")
    print(f"Maximum: {df[column].max()}")
    print(f"25% Quartil: {df[column].quantile(0.25):.2f}")
    print(f"75% Quartil: {df[column].quantile(0.75):.2f}")
    print(f"IQR: {df[column].quantile(0.75) - df[column].quantile(0.25):.2f}")
    print(f"Schiefe (Skewness): {df[column].skew():.2f}")
    print(f"Kurtosis: {df[column].kurtosis():.2f}")
    
    # Visualisierungen
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # Histogramm
    sns.histplot(df[column], bins=50, kde=True, ax=ax1)
    ax1.set_title(f'Histogramm von {column}')
    ax1.set_xlabel(column)
    ax1.set_ylabel('Häufigkeit')
    
    # Boxplot
    sns.boxplot(y=df[column], ax=ax2)
    ax2.set_title(f'Boxplot von {column}')
    ax2.set_ylabel(column)
    
    plt.tight_layout()
    plt.show()
    
    # Erkenntnisse
    if abs(df[column].skew()) < 0.5:
        skew_desc = "symmetrisch"
    elif df[column].skew() > 0:
        skew_desc = "rechtsschief (rechter Ausreißer)"
    else:
        skew_desc = "linksschief (linker Ausreißer)"
    
    print(f"\nErkenntnisse:")
    print(f"- Die Verteilung ist {skew_desc}.")
    print(f"- Der IQR beträgt {df[column].quantile(0.75) - df[column].quantile(0.25):.2f}, was auf die Streuung der mittleren 50% der Daten hinweist.")
    if df[column].max() - df[column].quantile(0.75) > 1.5 * (df[column].quantile(0.75) - df[column].quantile(0.25)):
        print(f"- Es gibt potenzielle Ausreißer nach oben (Maximum: {df[column].max()}).")
    if df[column].quantile(0.25) - df[column].min() > 1.5 * (df[column].quantile(0.75) - df[column].quantile(0.25)):
        print(f"- Es gibt potenzielle Ausreißer nach unten (Minimum: {df[column].min()}).")

# Numerische Variablen
numeric_columns = ['age', 'balance', 'day', 'duration', 'campaign', 'pdays', 'previous']

for col in numeric_columns:
    analyze_numeric_variable(df, col)
```

## 3. Analyse kategorischer Variablen

Wir analysieren die kategorischen Variablen: job, marital, education, default, housing, loan, contact, month, poutcome, y.

Für jede Variable berechnen wir:
- Häufigkeiten und Prozentsätze
- Visualisierungen (Barplots, Pie Charts)

```python
# Funktion für univariate Analyse kategorischer Variablen
def analyze_categorical_variable(df, column):
    print(f"\n=== Analyse von {column} ===")
    print(f"Anzahl der Werte: {len(df[column])}")
    print(f"Einzigartige Werte: {df[column].nunique()}")
    print(f"Fehlende Werte: {df[column].isnull().sum()}")
    print(f"Modus: {df[column].mode().iloc[0]}")
    
    # Häufigkeiten
    value_counts = df[column].value_counts()
    percentages = df[column].value_counts(normalize=True) * 100
    
    print(f"\nHäufigkeiten:")
    for val, count in value_counts.items():
        print(f"{val}: {count} ({percentages[val]:.2f}%)")
    
    # Visualisierungen
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # Barplot
    sns.barplot(x=value_counts.index, y=value_counts.values, ax=ax1, palette='viridis')
    ax1.set_title(f'Barplot von {column}')
    ax1.set_xlabel(column)
    ax1.set_ylabel('Häufigkeit')
    ax1.tick_params(axis='x', rotation=45)
    
    # Pie Chart (nur für wenige Kategorien)
    if len(value_counts) <= 10:
        ax2.pie(percentages, labels=percentages.index, autopct='%1.1f%%', startangle=90)
        ax2.set_title(f'Pie Chart von {column}')
        ax2.axis('equal')
    else:
        # Wenn zu viele Kategorien, zeige Top 10
        top10 = value_counts.head(10)
        top10_perc = (top10 / len(df)) * 100
        ax2.pie(top10_perc, labels=top10.index, autopct='%1.1f%%', startangle=90)
        ax2.set_title(f'Pie Chart Top 10 von {column}')
        ax2.axis('equal')
    
    plt.tight_layout()
    plt.show()
    
    # Erkenntnisse
    entropy = -sum((percentages/100) * np.log2(percentages/100))
    print(f"\nErkenntnisse:")
    print(f"- Die Variable hat {df[column].nunique()} einzigartige Werte.")
    print(f"- Die häufigste Kategorie ist '{df[column].mode().iloc[0]}' mit {value_counts.iloc[0]} Beobachtungen ({percentages.iloc[0]:.2f}%).")
    if entropy < 1:
        print("- Die Verteilung ist stark konzentriert (niedrige Entropie).")
    elif entropy < 2:
        print("- Die Verteilung ist moderat konzentriert.")
    else:
        print("- Die Verteilung ist breit gestreut (hohe Entropie).")

# Kategorische Variablen
categorical_columns = ['job', 'marital', 'education', 'default', 'housing', 'loan', 'contact', 'month', 'poutcome', 'y']

for col in categorical_columns:
    analyze_categorical_variable(df, col)
```

## 5. Variablentransformationen

Basierend auf der Analyse führen wir notwendige Transformationen durch, um die Daten für maschinelles Lernen vorzubereiten. Dies umfasst:

- Behandlung von Ausreißern
- Kodierung kategorischer Variablen
- Skalierung numerischer Variablen
- Behandlung des Klassenungleichgewichts
- Feature Engineering

```python
# Zusätzliche Imports für Transformationen
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline
import numpy as np
```

```python
# Kopie des DataFrames für Transformationen
df_transformed = df.copy()

print("Original DataFrame Shape:", df_transformed.shape)
print("Original Target Distribution:")
print(df_transformed['y'].value_counts(normalize=True))
```

```python
# 1. Behandlung von Ausreißern
# Log-Transformation für rechtsschiefe Variablen mit positiven Werten
df_transformed['balance_log'] = np.log1p(df_transformed['balance'] - df_transformed['balance'].min() + 1)  # Shift to positive
df_transformed['duration_log'] = np.log1p(df_transformed['duration'])
df_transformed['campaign_log'] = np.log1p(df_transformed['campaign'])

# Winsorizing für extreme Ausreißer (z.B. balance)
from scipy.stats.mstats import winsorize
df_transformed['balance_winsorized'] = winsorize(df_transformed['balance'], limits=[0.05, 0.05])

print("Ausreißer-Behandlung abgeschlossen.")
print("Neue Spalten hinzugefügt: balance_log, duration_log, campaign_log, balance_winsorized")
```

```python
# 2. Kodierung kategorischer Variablen
# Label Encoding für binäre Variablen
binary_vars = ['default', 'housing', 'loan', 'y']
le = LabelEncoder()
for var in binary_vars:
    df_transformed[f'{var}_encoded'] = le.fit_transform(df_transformed[var])

# One-Hot-Encoding für nominale Variablen
nominal_vars = ['job', 'marital', 'education', 'contact', 'poutcome']
df_transformed = pd.get_dummies(df_transformed, columns=nominal_vars, prefix=nominal_vars, drop_first=True)

# Ordinal Encoding für month (zeitliche Ordnung)
month_order = ['jan', 'feb', 'mar', 'apr', 'may', 'jun', 'jul', 'aug', 'sep', 'oct', 'nov', 'dec']
df_transformed['month_encoded'] = df_transformed['month'].map({month: i for i, month in enumerate(month_order)})

print("Kodierung abgeschlossen.")
print("Neue kodierte Spalten hinzugefügt.")
```

```python
# 3. Skalierung numerischer Variablen
numeric_features = ['age', 'balance_log', 'day', 'duration_log', 'campaign_log', 'pdays', 'previous', 'month_encoded']

scaler = StandardScaler()
df_transformed[[f'{col}_scaled' for col in numeric_features]] = scaler.fit_transform(df_transformed[numeric_features])

print("Skalierung abgeschlossen.")
print("Neue skalierte Spalten hinzugefügt.")
```

```python
# 4. Feature Engineering
# Neue Features erstellen
df_transformed['has_previous_contact'] = (df_transformed['pdays'] != -1).astype(int)
df_transformed['contact_intensity'] = df_transformed['campaign'] / (df_transformed['previous'] + 1)  # Vermeide Division durch 0
df_transformed['age_group'] = pd.cut(df_transformed['age'], bins=[0, 25, 35, 50, 65, 100], labels=['young', 'young_adult', 'adult', 'middle_aged', 'senior'])

# One-Hot für age_group
df_transformed = pd.get_dummies(df_transformed, columns=['age_group'], prefix='age_group', drop_first=True)

print("Feature Engineering abgeschlossen.")
print("Neue Features: has_previous_contact, contact_intensity, age_group (one-hot encoded)")
```

```python
# 5. Behandlung des Klassenungleichgewichts
# Auswahl der Features für das Modell
feature_cols = [col for col in df_transformed.columns if col.endswith(('_scaled', '_encoded')) or col.startswith(('job_', 'marital_', 'education_', 'contact_', 'poutcome_', 'age_group_', 'has_previous_contact', 'contact_intensity'))]
feature_cols += ['balance_winsorized']  # Nicht-skalierte Version für SMOTE

X = df_transformed[feature_cols]
y = df_transformed['y_encoded']

print("Features für Modell:", len(feature_cols))
print("Feature Namen:", feature_cols[:10], "...")  # Erste 10 anzeigen

# SMOTE für Oversampling der Minority-Klasse
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)

print("Nach SMOTE:")
print("Original y distribution:", y.value_counts().to_dict())
print("Resampled y distribution:", pd.Series(y_resampled).value_counts().to_dict())

# Finaler transformierter DataFrame
df_final = pd.DataFrame(X_resampled, columns=feature_cols)
df_final['y_encoded'] = y_resampled

print("Finaler DataFrame Shape:", df_final.shape)
```

```python
# Zusammenfassung der Transformationen
print("=== Zusammenfassung der Variablentransformationen ===")
print(f"Original Datensatz: {df.shape[0]} Zeilen, {df.shape[1]} Spalten")
print(f"Transformierter Datensatz: {df_transformed.shape[0]} Zeilen, {df_transformed.shape[1]} Spalten")
print(f"Finaler Datensatz nach SMOTE: {df_final.shape[0]} Zeilen, {df_final.shape[1]} Spalten")
print(f"Features für ML-Modell: {len(feature_cols)}")
print("\nDurchgeführte Transformationen:")
print("1. Ausreißer-Behandlung: Log-Transformation und Winsorizing")
print("2. Kodierung: Label-Encoding für binäre, One-Hot für nominale, Ordinal für zeitliche Variablen")
print("3. Skalierung: StandardScaler für numerische Features")
print("4. Feature Engineering: Neue Features wie has_previous_contact, contact_intensity, age_group")
print("5. Klassenbalance: SMOTE für Oversampling der Minority-Klasse")

# Speichere den finalen Datensatz
df_final.to_csv('bank-full-transformed.csv', index=False)
print("\nFinaler Datensatz gespeichert als 'bank-full-transformed.csv'")
```

## 6. Bivariate und multivariate Analysen

In diesem Abschnitt analysieren wir die Zusammenhänge zwischen verschiedenen Variablen, insbesondere im Hinblick auf die Zielvariable `y` (Erfolg der Marketingkampagne).

```python
# Korrelationsmatrix für numerische Variablen
numeric_cols = ['age', 'balance', 'day', 'duration', 'campaign', 'pdays', 'previous']

plt.figure(figsize=(10, 8))
correlation_matrix = df[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Korrelationsmatrix numerischer Variablen')
plt.show()

print("Stärkste Korrelationen:")
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        corr = correlation_matrix.iloc[i, j]
        if abs(corr) > 0.3:  # Nur signifikante Korrelationen
            print(f"{correlation_matrix.columns[i]} vs {correlation_matrix.columns[j]}: {corr:.3f}")
```

```python
# Kreuztabellen: Kategorische Variablen vs. Zielvariable y
categorical_vs_y = ['job', 'marital', 'education', 'default', 'housing', 'loan', 'contact', 'month', 'poutcome']

fig, axes = plt.subplots(3, 3, figsize=(18, 15))
axes = axes.flatten()

for i, cat_var in enumerate(categorical_vs_y):
    if i < len(axes):
        # Kreuztabelle normalisiert
        cross_tab = pd.crosstab(df[cat_var], df['y'], normalize='index')
        cross_tab.plot(kind='bar', stacked=True, ax=axes[i], colormap='viridis')
        axes[i].set_title(f'{cat_var} vs. y (Konversionsrate)')
        axes[i].set_ylabel('Anteil')
        axes[i].tick_params(axis='x', rotation=45)
        axes[i].legend(title='y', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
# Boxplots: Numerische Variablen vs. Zielvariable y
numeric_vs_y = ['age', 'balance', 'duration', 'campaign']

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.flatten()

for i, num_var in enumerate(numeric_vs_y):
    sns.boxplot(x='y', y=num_var, data=df, ax=axes[i], palette='Set2')
    axes[i].set_title(f'{num_var} vs. y')
    axes[i].set_xlabel('y (Erfolg)')
    axes[i].set_ylabel(num_var)

plt.tight_layout()
plt.show()

# Statistiken für Unterschiede
print("Mittelwert-Unterschiede zwischen y=no und y=yes:")
for var in numeric_vs_y:
    mean_no = df[df['y'] == 'no'][var].mean()
    mean_yes = df[df['y'] == 'yes'][var].mean()
    print(f"{var}: no={mean_no:.2f}, yes={mean_yes:.2f}, Differenz={mean_yes-mean_no:.2f}")
```

```python
# Scatterplots für wichtige numerische Beziehungen
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
axes = axes.flatten()

# Balance vs Age
sns.scatterplot(x='age', y='balance', hue='y', data=df, ax=axes[0], alpha=0.6)
axes[0].set_title('Balance vs. Age (nach y gefärbt)')

# Duration vs Balance
sns.scatterplot(x='balance', y='duration', hue='y', data=df, ax=axes[1], alpha=0.6)
axes[1].set_title('Duration vs. Balance (nach y gefärbt)')

# Campaign vs Previous
sns.scatterplot(x='previous', y='campaign', hue='y', data=df, ax=axes[2], alpha=0.6)
axes[2].set_title('Campaign vs. Previous (nach y gefärbt)')

# Age vs Duration
sns.scatterplot(x='age', y='duration', hue='y', data=df, ax=axes[3], alpha=0.6)
axes[3].set_title('Age vs. Duration (nach y gefärbt)')

plt.tight_layout()
plt.show()
```

```python
# Heatmap für kategorische Kreuztabellen (z.B. Job vs Education)
plt.figure(figsize=(12, 8))
job_education_cross = pd.crosstab(df['job'], df['education'], normalize='index')
sns.heatmap(job_education_cross, annot=True, fmt='.2f', cmap='YlGnBu')
plt.title('Heatmap: Job vs. Education (normalisiert nach Job)')
plt.xlabel('Education')
plt.ylabel('Job')
plt.show()

# Pairplot für wichtige numerische Variablen (Sample für Performance)
sample_df = df.sample(1000, random_state=42)  # Sample für schnellere Darstellung
important_numeric = ['age', 'balance', 'duration', 'campaign']
sns.pairplot(sample_df[important_numeric + ['y']], hue='y', diag_kind='kde', plot_kws={'alpha': 0.6})
plt.suptitle('Pairplot wichtiger numerischer Variablen (Sample)', y=1.02)
plt.show()
```
