# Univariate Analyse des Bank Marketing Datasets

Dieses Notebook führt eine umfassende univariate Analyse der Daten aus der `bank-full.csv` durch. Wir analysieren jede Variable einzeln, um ihre Verteilung, Statistiken und potenzielle Erkenntnisse zu verstehen.

## Inhaltsverzeichnis
1. Daten laden und Überblick
2. Analyse numerischer Variablen
3. Analyse kategorischer Variablen
4. Schlussfolgerungen

```python
# Import erforderlicher Bibliotheken
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Einstellungen für Plots
plt.style.use('seaborn-v0_8')
sns.set_palette('husl')
plt.rcParams['figure.figsize'] = (10, 6)

# Warnungen unterdrücken
import warnings
warnings.filterwarnings('ignore')
```

## 1. Daten laden und Überblick

Laden wir die Daten und verschaffen uns einen ersten Überblick über die Struktur und die grundlegenden Statistiken.

```python
# Daten laden
df = pd.read_csv('bank-full.csv', sep=';')

# Grundlegender Überblick
print("Form des Datensatzes:", df.shape)
print("\nSpaltennamen:")
print(df.columns.tolist())
print("\nDatentypen:")
print(df.dtypes)
print("\nErste 5 Zeilen:")
display(df.head())
print("\nZusammenfassung der numerischen Variablen:")
display(df.describe())
print("\nZusammenfassung der kategorischen Variablen:")
display(df.describe(include=['object']))
print("\nFehlende Werte:")
print(df.isnull().sum())
```

## 2. Analyse numerischer Variablen

Wir analysieren die numerischen Variablen: age, balance, day, duration, campaign, pdays, previous.

Für jede Variable berechnen wir:
- Grundlegende Statistiken
- Verteilungsmaße (Schiefe, Kurtosis)
- Visualisierungen (Histogramme, Boxplots)

```python
# Funktion für univariate Analyse numerischer Variablen
def analyze_numeric_variable(df, column):
    print(f"\n=== Analyse von {column} ===")
    print(f"Anzahl der Werte: {len(df[column])}")
    print(f"Einzigartige Werte: {df[column].nunique()}")
    print(f"Fehlende Werte: {df[column].isnull().sum()}")
    print(f"Mittelwert: {df[column].mean():.2f}")
    print(f"Median: {df[column].median():.2f}")
    print(f"Modus: {df[column].mode().iloc[0]}")
    print(f"Standardabweichung: {df[column].std():.2f}")
    print(f"Minimum: {df[column].min()}")
    print(f"Maximum: {df[column].max()}")
    print(f"25% Quartil: {df[column].quantile(0.25):.2f}")
    print(f"75% Quartil: {df[column].quantile(0.75):.2f}")
    print(f"IQR: {df[column].quantile(0.75) - df[column].quantile(0.25):.2f}")
    print(f"Schiefe (Skewness): {df[column].skew():.2f}")
    print(f"Kurtosis: {df[column].kurtosis():.2f}")
    
    # Visualisierungen
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # Histogramm
    sns.histplot(df[column], bins=50, kde=True, ax=ax1)
    ax1.set_title(f'Histogramm von {column}')
    ax1.set_xlabel(column)
    ax1.set_ylabel('Häufigkeit')
    
    # Boxplot
    sns.boxplot(y=df[column], ax=ax2)
    ax2.set_title(f'Boxplot von {column}')
    ax2.set_ylabel(column)
    
    plt.tight_layout()
    plt.show()
    
    # Erkenntnisse
    if abs(df[column].skew()) < 0.5:
        skew_desc = "symmetrisch"
    elif df[column].skew() > 0:
        skew_desc = "rechtsschief (rechter Ausreißer)"
    else:
        skew_desc = "linksschief (linker Ausreißer)"
    
    print(f"\nErkenntnisse:")
    print(f"- Die Verteilung ist {skew_desc}.")
    print(f"- Der IQR beträgt {df[column].quantile(0.75) - df[column].quantile(0.25):.2f}, was auf die Streuung der mittleren 50% der Daten hinweist.")
    if df[column].max() - df[column].quantile(0.75) > 1.5 * (df[column].quantile(0.75) - df[column].quantile(0.25)):
        print(f"- Es gibt potenzielle Ausreißer nach oben (Maximum: {df[column].max()}).")
    if df[column].quantile(0.25) - df[column].min() > 1.5 * (df[column].quantile(0.75) - df[column].quantile(0.25)):
        print(f"- Es gibt potenzielle Ausreißer nach unten (Minimum: {df[column].min()}).")

# Numerische Variablen
numeric_columns = ['age', 'balance', 'day', 'duration', 'campaign', 'pdays', 'previous']

for col in numeric_columns:
    analyze_numeric_variable(df, col)
```

## 3. Analyse kategorischer Variablen

Wir analysieren die kategorischen Variablen: job, marital, education, default, housing, loan, contact, month, poutcome, y.

Für jede Variable berechnen wir:
- Häufigkeiten und Prozentsätze
- Visualisierungen (Barplots, Pie Charts)

```python
# Funktion für univariate Analyse kategorischer Variablen
def analyze_categorical_variable(df, column):
    print(f"\n=== Analyse von {column} ===")
    print(f"Anzahl der Werte: {len(df[column])}")
    print(f"Einzigartige Werte: {df[column].nunique()}")
    print(f"Fehlende Werte: {df[column].isnull().sum()}")
    print(f"Modus: {df[column].mode().iloc[0]}")
    
    # Häufigkeiten
    value_counts = df[column].value_counts()
    percentages = df[column].value_counts(normalize=True) * 100
    
    print(f"\nHäufigkeiten:")
    for val, count in value_counts.items():
        print(f"{val}: {count} ({percentages[val]:.2f}%)")
    
    # Visualisierungen
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # Barplot
    sns.barplot(x=value_counts.index, y=value_counts.values, ax=ax1, palette='viridis')
    ax1.set_title(f'Barplot von {column}')
    ax1.set_xlabel(column)
    ax1.set_ylabel('Häufigkeit')
    ax1.tick_params(axis='x', rotation=45)
    
    # Pie Chart (nur für wenige Kategorien)
    if len(value_counts) <= 10:
        ax2.pie(percentages, labels=percentages.index, autopct='%1.1f%%', startangle=90)
        ax2.set_title(f'Pie Chart von {column}')
        ax2.axis('equal')
    else:
        # Wenn zu viele Kategorien, zeige Top 10
        top10 = value_counts.head(10)
        top10_perc = (top10 / len(df)) * 100
        ax2.pie(top10_perc, labels=top10.index, autopct='%1.1f%%', startangle=90)
        ax2.set_title(f'Pie Chart Top 10 von {column}')
        ax2.axis('equal')
    
    plt.tight_layout()
    plt.show()
    
    # Erkenntnisse
    entropy = -sum((percentages/100) * np.log2(percentages/100))
    print(f"\nErkenntnisse:")
    print(f"- Die Variable hat {df[column].nunique()} einzigartige Werte.")
    print(f"- Die häufigste Kategorie ist '{df[column].mode().iloc[0]}' mit {value_counts.iloc[0]} Beobachtungen ({percentages.iloc[0]:.2f}%).")
    if entropy < 1:
        print("- Die Verteilung ist stark konzentriert (niedrige Entropie).")
    elif entropy < 2:
        print("- Die Verteilung ist moderat konzentriert.")
    else:
        print("- Die Verteilung ist breit gestreut (hohe Entropie).")

# Kategorische Variablen
categorical_columns = ['job', 'marital', 'education', 'default', 'housing', 'loan', 'contact', 'month', 'poutcome', 'y']

for col in categorical_columns:
    analyze_categorical_variable(df, col)
```

## 4. Schlussfolgerungen

### Zusammenfassung der wichtigsten Erkenntnisse:

**Numerische Variablen:**
- **Age**: Rechtsschiefe Verteilung, Durchschnittsalter ~41 Jahre, Ausreißer bis 95.
- **Balance**: Stark rechtsschiefe Verteilung mit vielen negativen Werten und hohen Ausreißern (bis 102k).
- **Day**: Symmetrische Verteilung über die Monatstage.
- **Duration**: Stark rechtsschiefe Verteilung, wichtige Variable für Erfolg (längere Gespräche = höhere Konversionsrate).
- **Campaign**: Rechtsschiefe, meist 1-3 Kontakte, Ausreißer bis 63.
- **Pdays/Previous**: Viele -1 (keine vorherigen Kontakte), rechtsschiefe für vorherige Kontakte.

**Kategorische Variablen:**
- **Job**: Blue-collar am häufigsten (21.5%), gefolgt von Management.
- **Marital**: 60% verheiratet, 28% single.
- **Education**: 51% secondary, 29% tertiary.
- **Default/Housing/Loan**: Meist "no", stark unbalanciert.
- **Contact**: 65% cellular, 29% unknown.
- **Month**: Mai am häufigsten (30%), saisonale Muster.
- **Poutcome**: 82% unknown, zeigt Datenqualitätsproblem.
- **Y (Target)**: Stark unbalanciert, nur 11.7% "yes".
