# Univariate Analyse des Bank Marketing Datasets
Dieses Notebook führt eine univariate Analyse der Daten aus der bank-full.csv durch. 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings

plt.style.use('seaborn-v0_8')
sns.set_palette('husl')
plt.rcParams['figure.figsize'] = (10, 6)
warnings.filterwarnings('ignore')
```

```python
raw_data = pd.read_csv('bank-full.csv', sep=';')

print(raw_data.shape)
print(raw_data.columns.tolist())
display(raw_data.describe())
display(raw_data.describe(include=['object']))
```

```python
print(f"Anzahl der Werte: {len(raw_data['age'])}")
print(f"Einzigartige Werte: {raw_data['age'].nunique()}")
print(f"Fehlende Werte: {raw_data['age'].isnull().sum()}")
print(f"Mittelwert: {raw_data['age'].mean():.2f}")
print(f"Median: {raw_data['age'].median():.2f}")
print(f"Minimum: {raw_data['age'].min()}")
print(f"Maximum: {raw_data['age'].max()}")
print(f"25% Quartil: {raw_data['age'].quantile(0.25):.2f}")
print(f"75% Quartil: {raw_data['age'].quantile(0.75):.2f}")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

sns.histplot(raw_data['age'], bins=50, kde=True, ax=ax1)
ax1.set_title(f'Histogram of age')
ax1.set_xlabel('age')
ax1.set_ylabel('Frequency')

sns.boxplot(y=raw_data['age'], ax=ax2)
ax2.set_title(f'Boxplot of age')
ax2.set_ylabel('age')
```

# Variablentranformation
Umwandeln der Variablen in eine für ML geeignete Form. Handling von Ausreißern und Missing Values

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline

data_transformed = raw_data.copy()
```

## Age
Missing: keine
Außreißer: Ja durch wenige Alte Personen, sind aber korrekt deswegen behalten

Umformung: Binning in Altersgruppen, One-Hot Encoding

```python
data_transformed['age_group'] = pd.cut(data_transformed['age'], bins=[0, 25, 35, 50, 65, 100], labels=['young', 'young_adult', 'adult', 'middle_aged', 'senior'])
data_transformed = pd.get_dummies(data_transformed, columns=['age_group'], prefix='age_group', drop_first=True)
data_transformed = data_transformed.drop('age', axis=1)
```

## Job
Missing: 288 Unknown
Ausreißer: keine

Umformung:
Unknown als eigene Kategorie behalten, Hot-One Encoding

```python
data_transformed = pd.get_dummies(data_transformed, columns=['job'], prefix='job', drop_first=True)
```

## Marital
Missing: keine
Ausreißer: keine

Umformung: keine Änderungen nötig, Hot-One Encoding

## Aufteilen des Datensatztes
Der Datensatzt unterteilt sich in Einträge bei denen der Kunde in früheren Kampanien bereits kontaktiert wurde und in Kunden die zum ersten mal kontaktiert werden.

Beide Teile müssen getrennt und einzelnd betrachtet werden!

```python
data_firsts_final = data_transformed[data_transformed['pdays'] == -1]
data_previous_final = data_transformed[data_transformed['pdays'] != -1]

data_firsts_final = data_firsts_final.drop('pdays', axis=1)
data_firsts_final = data_firsts_final.drop('previous', axis=1)
data_firsts_final = data_firsts_final.drop('poutcome', axis=1)
```
